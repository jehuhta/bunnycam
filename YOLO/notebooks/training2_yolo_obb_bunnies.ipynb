## Training

```python
# Initialize our yolo model as 'model'
model = YOLO("yolo11n.pt")

# Trains the model using the yaml instructions.
train = model.train(
    data=r"C:\Users\kykis\bunnycam\YOLO\datasets\model_2\dataset.yaml",   # path to the yaml file
    epochs=30,                                                            # Amount of batch cycles
    imgsz=1024,                                                           # Image size
    workers=2,                                                            # Reduce this to lower ram usage amount.
    device=0                                                              # Use GPU
)
```
