### Autosplitter 

<i> This notebook splits image labels for us.

```python
# Imports
from ultralytics.data.split import autosplit
```

```python
# Splits the images based on the path.
autosplit(
    path=r"C:\Users\kykis\bunnycam\YOLO\datasets\model_2\images",
    weights=(0.9, 0.1, 0.0),             # (train, validation, test) fractional splits
    annotated_only=False,                # split only images with annotation file when True
)
```

<i>
This will create autosplit_*.txt files. Then use txt-files in training like this:
path: POINT TO THE Autosplit -folder
train: autosplit_train.txt
val: autosplit_val.txt
test: autosplit_test.txt
