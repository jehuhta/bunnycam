### Autosplitter
##### Dataset 3 

<i> This notebook splits image labels for us.

```python
# Imports
from ultralytics.data.split import autosplit
```

```python
# Splits the images based on the path.
autosplit(
    path=r"C:\Users\kykis\bunnycam\YOLO\datasets\model_5\images",
    weights=(0.8, 0.1, 0.1),             # (train, validation, test) fractional splits
    annotated_only=True,                # split only images with annotation file when True
)

# Autosplit shuffles automatically ('autosplit uses random.seed(0) for reproducibility,
# then calls random.choices([0, 1, 2], weights=weights, k=n)
# to randomly assign each file to train, val, or test.')
```

<i>
This will create autosplit_*.txt files. Then use txt-files in training like this:
path: POINT TO THE Autosplit -folder
train: autosplit_train.txt
val: autosplit_val.txt
test: autosplit_test.txt
