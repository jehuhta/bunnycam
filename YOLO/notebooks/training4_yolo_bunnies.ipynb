## Imports

--- 

## Training

```python
# Initialize our yolo model as 'model'
model = YOLO("yolo11s.pt")

# Trains the model using the yaml instructions.
# Data augmentation lines have been added to improve the model.
train = model.train(
    data=r"C:\Users\kykis\bunnycam\YOLO\datasets\model_4\dataset.yaml",   # path to the yaml file
    epochs=100,                                                            # Amount of batch cycles
    patience=15,                                                           # If it doesn't improve after 5 epochs, stop
    imgsz=640,                                                            # Image size
    workers=0,                                                            # Reduce this to lower ram usage amount.
    device=0,                                                             # Use GPU
    hsv_h=0.025,                                                          # hue shift
    hsv_s=.5,                                                             # saturation                                                         
    hsv_v=.5,                                                             # brightness
    flipud=0.1,                                                           # Upside down flip chance
    fliplr=0.5,                                                           # horizontal flip probability
    degrees=18.0,                                                         # rotation
    scale=0.5,                                                            # zoom
    translate=0.2,                                                        # shift
    shear=2.0,
)
```
