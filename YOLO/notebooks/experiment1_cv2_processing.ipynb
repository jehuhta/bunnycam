```python
import cv2
import numpy as np

img = cv2.imread(r"C:\Users\kykis\bunnycam\YOLO\datasets\validation_2\device_E8F60A870078_date_13-05-26_hours_01.54.53.jpg")

# Denoise first, then CLAHE
denoised = cv2.fastNlMeansDenoisingColored(img, None, 
    h=1,           # filter strength, increase if still noisy
    hColor=30,      # color filter strength
    templateWindowSize=7, 
    searchWindowSize=21
)

lab = cv2.cvtColor(denoised, cv2.COLOR_BGR2LAB)
l, a, b = cv2.split(lab)
clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
l = clahe.apply(l)
enhanced = cv2.merge((l, a, b))
enhanced = cv2.cvtColor(enhanced, cv2.COLOR_LAB2BGR)

cv2.imwrite("enhanced.jpg", enhanced)
```
