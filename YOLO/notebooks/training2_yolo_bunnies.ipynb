## Training

```python
# Initialize our yolo model as 'model'
model = YOLO("yolo11s.pt")

# Trains the model using the yaml instructions.
train = model.train(
    data=r"C:\Users\kykis\bunnycam\YOLO\datasets\model_2\dataset.yaml",   # path to the yaml file
    epochs=20,                                                            # Amount of batch cycles
    imgsz=640,                                                            # Image size
    workers=0,                                                            # Reduce this to lower ram usage amount.
    device=0                                                              # Use GPU
)
```
