```python
import duckdb
```

#### Creating the database

```python
# Creating the database and connecting to it at the same time.
con = duckdb.connect("bunnycam.db")
```

#### Creating a table for bunny predictions

```python
# Let's create a table with a timestamp some columns with their appropriate datatypes. 
con.sql("""

CREATE TABLE bunny_predictions (
    Timestamp TIMESTAMP_S,
    Dippi_Predicted FLOAT,
    Aune_Predicted FLOAT,
    Dippi_Score FLOAT,
    Aune_Score FLOAT;
)

""")
```

#### Tester Row

Creating the values

```python
# Let's create a tester row to make sure this TABLE is functioning as intended.
from datetime import datetime

# Get today's time
now = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
print(now)

# Add the date with some test values.
con.sql(f"""

INSERT INTO bunny_predictions VALUES (
'{now}',.1,1,.5,1
)

""")
```

Checking the table to see if it worked

```python
con.sql(f"""

SELECT * FROM bunny_predictions

""")
```
